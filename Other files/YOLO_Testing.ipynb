```python
from ultralytics import YOLO

model = YOLO("vegetable_detector_yolov8n.pt")
```

```python
for i in range(1,4):
  results = model.predict(
      source=f"test{i}.jpg",
      conf=0.3,
      save=True
      )
```

```python
results = model.predict(
    source=f"test1.jpg",
    conf=0.3,
    save=True
    )
```

```python
detected = []

for r in results:
    for box in r.boxes:
        cls_id = int(box.cls[0])
        conf = float(box.conf[0])
        name = model.names[cls_id]
        detected.append((name, conf))

print(detected)
```
